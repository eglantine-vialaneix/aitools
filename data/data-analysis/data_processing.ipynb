```python
import pandas as pd
```

```python
df = pd.read_json("../experiment-collections/collection_pilot_2026-05-26.jsonl", lines=True)
df
```

```python
df["DLDuration (min)"] = round((pd.to_datetime(df["DLEndTime"]) - pd.to_datetime(df["DLStartTime"])).dt.total_seconds() / 60, 1)
df["FSDuration (min)"] = round((pd.to_datetime(df["FSEndTime"]) - pd.to_datetime(df["FSStartTime"])).dt.total_seconds() / 60, 1)
df["ModelDuration (min)"] = round((pd.to_datetime(df["ModelEndTime"]) - pd.to_datetime(df["ModelStartTime"])).dt.total_seconds() / 60, 1)
df["EvalDuration (min)"] = round((pd.to_datetime(df["EvalEndTime"]) - pd.to_datetime(df["EvalStartTime"])).dt.total_seconds() / 60, 1)
```
