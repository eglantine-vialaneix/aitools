```python
import pandas as pd
import numpy as np
```

```python
path_to_scraped_dataset = "../../public/Dino Datasets Scraped.csv"
full_df = pd.read_csv(path_to_scraped_dataset, sep=';')
full_df
```

```python
full_df.columns
```

```python
# we drop 'taxonomy', 'link', 'Story', 'dentition',
clean_en_df = full_df[['name', 'diet', 'period', 'lived_in', 'type', 'bipedal', 'length',
       'weight', 'named_by', 'species', 'taxonomy_suborder', 'taxonomy_family']]
clean_en_df
```

```python
# Translating column names

clean_fr_df = clean_en_df.copy()
column_translation = {
    "name": "nom",
    "diet": "régime_alimentaire",
    "period": "période",
    "lived_in": "habitat",
    "type": "type",
    "bipedal": "bipède",
    "length": "longueur (m)",
    "weight": "poids (kg)",
    "named_by": "nommé_par",
    "species": "espèce",
    "taxonomy_suborder": "sous-ordre_taxonomique",
    "taxonomy_family": "famille_taxonomique"
}
clean_fr_df = clean_fr_df.rename(columns=column_translation)
clean_fr_df.head()
```

```python
# Cleaning periods in simpler eras
clean_fr_df["période"] = clean_fr_df["période"].apply(lambda x: ' '.join(x.replace(',', '').split()[:2])) 
clean_fr_df["période"]
```

```python
# Removing species null values
clean_fr_df = clean_fr_df[~clean_fr_df["espèce"].isnull()]

# Cleaning one weird line where the species starts with a '\n'
bad_index = clean_fr_df[clean_fr_df["espèce"].apply(lambda x:'\n' in x)].index
display(clean_fr_df[clean_fr_df["espèce"].apply(lambda x:'\n' in x)])
print("bad index:", bad_index)
clean_fr_df.loc[bad_index, "espèce"] = clean_fr_df.loc[bad_index, "espèce"].str.replace('\n', '')
clean_fr_df.loc[bad_index, "espèce"]
```

```python
# BIG MAPPER
# For each column name we map the translation of all elements
total_mapper = {"régime_alimentaire":  {'carnivorous':'carnivore',
                                        'herbivorous':'herbivore'},
                "période": {
                           "Late Cretaceous": "Crétacé supérieur",
                           "Late Jurassic": "Jurassique supérieur",
                           "Early Cretaceous": "Crétacé inférieur",
                           "Mid Jurassic": "Jurassique moyen",
                           "Early Jurassic": "Jurassique inférieur",
                           "Late Triassic": "Trias supérieur"
                           },
                "habitat": {
                           "USA": "États-Unis",
                           "Mongolia": "Mongolie",
                           "Morocco": "Maroc",
                           "Argentina": "Argentine",
                           "United Kingdom": "Royaume-Uni",
                           "UK": "Royaume-Uni",
                           "Germany": "Allemagne",
                           "Canada": "Canada",
                           "North Africa": "Afrique du Nord",
                           "Tanzania": "Tanzanie",
                           "Switzerland": "Suisse",
                           "Russia": "Russie",
                           "South Africa": "Afrique du Sud",
                           "China": "Chine",
                           "Australia": "Australie"
                           },
                "type": {
                        "large theropod": "grand théropode",
                        "small theropod": "petit théropode",
                        "ceratopsian": "cératopsien",
                        "sauropod": "sauropode",
                        "armoured dinosaur": "dinosaure cuirassé",
                        "euornithopod": "euornithopode",
                        "prosauropod": "prosauropode",
                        "small ornithischian": "petit ornithischien",
                        "pachycephalosaur": "pachycéphalosaure",
                        "small ornithopod": "petit ornithopode",
                        "large ornithopod": "grand ornithopode"
                        },
                "bipède": {
                           "Yes": True,
                           "No": False,
                           "flies": True #only one dino, and it actually has only 2 legs
                        }
}

# And we perform the translation
for key, mapper in total_mapper.items():
    clean_fr_df[key] = clean_fr_df[key].map(mapper).fillna(clean_fr_df[key])
```

```python
# Capitalize the first letter of the name
clean_fr_df["nom"] = clean_fr_df["nom"].apply(lambda x: x[0].upper() + x[1:])
```

```python
display(clean_fr_df[clean_fr_df["poids (kg)"].apply(lambda x: x[-1] == "m")])
bad_index = clean_fr_df[clean_fr_df["poids (kg)"].apply(lambda x: x[-1] == "m")].index
display(clean_fr_df.loc[bad_index, "poids (kg)"]) # NOT A WEIGHT
clean_fr_df.loc[bad_index, "poids (kg)"] = "75kg" #source: https://en.wikipedia.org/wiki/Deinonychus#Description
clean_fr_df.loc[bad_index, "poids (kg)"]
```

```python
# Cleaning longuer and poids to correct format
clean_fr_df["longueur (m)"] = clean_fr_df["longueur (m)"].apply(lambda x: float(x[:-1]))
clean_fr_df["poids (kg)"] = clean_fr_df["poids (kg)"].apply(lambda x: int(x[:-2].replace(' ', '')))
```

```python
display(clean_fr_df.loc[clean_fr_df["nom"]=="Gallimimus", "nommé_par"])
clean_fr_df.loc[clean_fr_df["nom"]=="Gallimimus", "nommé_par"] = "Osmòlska Roniewicz and Barsbold (1972)"
display(clean_fr_df.loc[clean_fr_df["nom"]=="Gallimimus", "nommé_par"])
```

```python
clean_fr_df["sous-ordre_taxonomique"] = clean_fr_df["sous-ordre_taxonomique"].apply(lambda x: x.replace(',', ''))
```
